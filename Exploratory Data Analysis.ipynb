```python
def conv(value):
    if value == 1:
        return "yes"
    else:
        return "no"

df['SeniorCitizen'] = df["SeniorCitizen"].apply(conv)
```

#converted 0 and 1 values of senior citizen to yes/no to make it easier to understand

```python
ax = sns.countplot(x = 'Churn', data = df)
ax.bar_label(ax.containers[0])
plt.title("count of customer by Churn")
plt.show
```

```python
gb = df.groupby("Churn").agg({'Churn':"count"})
plt.pie(gb['Churn'], labels = gb.index, autopct = "%1.2f%%")
plt.title("percentage of churned customers")

plt.show()
```

# from the given pie chart we can conclude that 26.54% of our customer have churned out. let's explore the reason behind it.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Percentage calculation
churn_pct = pd.crosstab(
    df['SeniorCitizen'],
    df['Churn'],
    normalize='index'
) * 100

# Stacked bar chart
ax = churn_pct.plot(
    kind='bar',
    stacked=True,
    figsize=(8, 5)
)

plt.title('Churn by SeniorCitizen (%)')
plt.xlabel('SeniorCitizen')
plt.ylabel('Percentage')
plt.xticks(rotation=0)
plt.legend(title='Churn')

# Percentage labels
for container in ax.containers:
    labels = [f'{v:.1f}%' if v > 0 else '' for v in container.datavalues]
    ax.bar_label(container, labels=labels, label_type='center')

plt.ylim(0, 100)
plt.show()
```

```python
columns = [
    'PhoneService',
    'MultipleLines',
    'InternetService',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies'
]

fig, axes = plt.subplots(3, 3, figsize=(18, 14))

for ax, col in zip(axes.flatten(), columns):
    sns.countplot(
        x=col,
        data=df,
        hue="Churn",
        ax=ax
    )

    ax.set_title(f'{col} vs Churn')
    ax.set_xlabel('')
    ax.set_ylabel('Count')
    ax.tick_params(axis='x', rotation=30)

    for container in ax.containers:
        ax.bar_label(container, fontsize=8)

plt.tight_layout()
plt.show()
```

#Customers with PhoneService, MultipleLines, and InternetService generally show more Churn = No than Churn = Yes.

OnlineSecurity, OnlineBackup, DeviceProtection, and TechSupport users have noticeably lower churn compared with customers who don't use these services.

StreamingTV and StreamingMovies also show higher numbers of non-churned customers, although churn is still present.

Overall, customers without additional security/support services tend to have a higher proportion of churn, suggesting these services may be associated with better customer retention.
